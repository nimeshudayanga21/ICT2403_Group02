```python
import cv2
import os
import matplotlib.pyplot as plt

def histogram(image, title):
    hist = cv2.calcHist([image], [0], None, [256], [0,256])
    plt.title(title)
    plt.xlabel("Intensity values")
    plt.ylabel("Frequency")
    plt.plot(hist)
    plt.xlim([0,256])


# Input folder
input_folder = r"C:\Users\Nimesh\Downloads\ICT2403_New\Frames"
# Output folder
output_folder = r"C:\Users\Nimesh\Downloads\ICT2403_New\Noice_Removed_Frames"

os.makedirs(output_folder, exist_ok=True)

image_extensions = ('.jpg', '.jpeg', '.png')

for filename in os.listdir(input_folder):

    if filename.lower().endswith(image_extensions):

        path = os.path.join(input_folder, filename)
        img = cv2.imread(path,0)

        if img is None:
            continue

        plt.figure(figsize=(16,8))

        plt.subplot(2,2,1)
        plt.imshow(img, cmap='gray')
        plt.title("original image")
        plt.axis('off')

        guassian = cv2.GaussianBlur(img,(5,5),1.0)

        plt.subplot(2,2,2)
        histogram(img,"original Histogram")

        plt.subplot(2,2,3)
        plt.imshow(guassian, cmap='gray')
        plt.title("guassian image")
        plt.axis('off')


        plt.subplot(2,2,4)
        histogram(guassian,"denoised")
        plt.tight_layout()
        plt.show()


        median = cv2.medianBlur(guassian,(5))
        plt.figure(figsize=(16,8))

        plt.subplot(2,2,1)
        plt.imshow(median, cmap='gray')
        plt.title("median image")
        plt.axis('off')

        plt.subplot(2,2,2)
        histogram(median,"denoised")

        cv2.imwrite(os.path.join(output_folder, f"{os.path.splitext(filename)[0]}_noise_removed.png"), median)
        plt.tight_layout()
        plt.show()

        print(img.shape)
```
