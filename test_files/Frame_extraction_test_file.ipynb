## Frame extraction

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# input video path
video_path = "Alligater_crack.mp4"
output_folder = "Frames"


cap = cv2.VideoCapture(video_path)

# check if video opened or not
if not cap.isOpened():
    print("Error opening video file")
    exit()

# calculation variales for frame extraction
frame_index = 0
saved_count = 0

# loop for read every frame
while True:
    # Read the next frame from the video; 'ret' is True or False, 'frame' is the image data
    ret, frame = cap.read()

    # if video end then loop will break
    if not ret:
        break

    # logic for frame extract
    # here we save a frame at every 10 frames
    if frame_index % 10 == 0:
        # Use 'f' for insert variables directly into the filename string
        filename = f"{output_folder}/frame_{saved_count}.jpg"
        cv2.imwrite(filename, frame)
        saved_count = saved_count + 1
    
    frame_index = frame_index + 1

cap.release()
print(f"Done! Saved {saved_count} frames.")
```
